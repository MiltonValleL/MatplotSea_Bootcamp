# LESSON 3: GRIDSPEC

---

>---
>You can build layouts with charts of varying sizes by setting a **gridspec** object
>
>- This creates a grid with a specified number of rows & columns
>
>- Each axis, or chart, can then occupy a group of squares in the grid
>
>---
>
>The starting configuration of the layout:
>
> | INITIAL CONFIGURATION |      | FINAL CONFIGURATION |
> |:---:|:---:|:---:|
> |  |      |  |
>
>
>---

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
```

```python
fig = plt.figure(figsize=(10, 5))
gs = gridspec.GridSpec(ncols=4, nrows=8)

ax1 = fig.add_subplot(gs[0:4, 0:2])
ax2 = fig.add_subplot(gs[0:4, 2:4])
ax3 = fig.add_subplot(gs[4:, :])

plt.tight_layout()
plt.show()
```

## LABORATORY

```python
df = pd.read_csv('../Data/Diamonds Prices2022.csv', usecols=list(range(1, 11)))
df.head()
```

```python
fig = plt.figure(figsize=(12, 6), constrained_layout=True)

grid = gridspec.GridSpec(8, 8, figure=fig)

ax1 = fig.add_subplot(grid[2:7, 0:6])
ax1.scatter(df.carat, df.price, alpha=0.3)

ax2 = fig.add_subplot(grid[0:2, 0:6])
ax2.hist(df.carat)

ax3 = fig.add_subplot(grid[2:7, 6:])
ax3.hist(df.price, orientation='horizontal')

plt.show()
```
